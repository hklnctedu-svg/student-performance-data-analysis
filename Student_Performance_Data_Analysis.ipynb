```python

import pandas as pd
import matplotlib.pyplot as plt

# 1. Create a sample dataset
data = {
    "Student": ["Aarav", "Diya", "Kabir", "Meera", "Rohan",
                "Ananya", "Vivaan", "Isha", "Arjun", "Sara"],
    "Python":  [85, 92, 67, 78, 55, 96, 72, 88, 45, 81],
    "Maths":   [78, 89, 70, 82, 60, 94, 68, 85, 52, 79],
    "DBMS":    [82, 95, 65, 80, 58, 91, 75, 87, 48, 84]
}

df = pd.DataFrame(data)

# 2. Calculate total and average marks
subjects = ["Python", "Maths", "DBMS"]

df["Total"] = df[subjects].sum(axis=1)
df["Average"] = df[subjects].mean(axis=1).round(2)

# 3. Assign result status
# Passing rule: at least 35 marks in every subject
df["Result"] = df[subjects].ge(35).all(axis=1).map(
    {True: "Pass", False: "Fail"}
)

# 4. Display the analysed dataset
print("STUDENT PERFORMANCE REPORT")
print(df.to_string(index=False))

# 5. Calculate subject-wise averages
subject_averages = df[subjects].mean().round(2)

print("\nSUBJECT-WISE AVERAGES")
print(subject_averages)

# 6. Find the highest average
top_student = df.loc[df["Average"].idxmax()]

print("\nTOP PERFORMER")
print("Student:", top_student["Student"])
print("Average:", top_student["Average"])

# 7. Summarize the results
print("\nCLASS SUMMARY")
print("Number of students:", len(df))
print("Students passed:", (df["Result"] == "Pass").sum())
print("Students failed:", (df["Result"] == "Fail").sum())
print("Class average:", round(df["Average"].mean(), 2))

# 8. Plot subject-wise average marks
subject_averages.plot(kind="bar")
plt.title("Average Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.ylim(0, 100)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# 9. Plot individual student averages
df.plot(
    x="Student",
    y="Average",
    kind="bar",
    legend=False
)
plt.title("Average Marks by Student")
plt.xlabel("Student")
plt.ylabel("Average Marks")
plt.ylim(0, 100)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 10. Save the report as a CSV file
df.to_csv("student_performance_report.csv", index=False)
print("\nReport saved as student_performance_report.csv")
```
